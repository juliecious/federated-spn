```python
log_weights = torch.log(w)

# Compute log(w_i) + log_values
log_terms = log_weights + a

# Use log-sum-exp trick for stability
max_log_term = torch.max(log_terms)  # max(log(w_i) + log_values)
log_sum = max_log_term + torch.log(torch.sum(torch.exp(log_terms - max_log_term)))
log_sum
```
